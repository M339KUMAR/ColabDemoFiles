#Task-4:  Matrix Operations

```python

%%writefile Matrix_Ops.py


import numpy as np

# -----------------------------------------
# Function to Input Matrix
# -----------------------------------------
def input_matrix(name):

    rows = int(input(f"Enter rows for {name}: "))
    cols = int(input(f"Enter columns for {name}: "))

    print(f"\nEnter elements of {name}")

    matrix = []

    for i in range(rows):

       # for j in range(cols):
       while True:
            row = list(map(float, input().split()))

            if len(row)== cols:
                matrix.append(row)
                break
            else:
                print(f"please enter exactly {cols} values for each row with a space in between:")

    return np.array(matrix)

# -----------------------------------------
# Display Menu
# -----------------------------------------
try:
 while True:

    print("\n==============================")
    print(" MATRIX OPERATIONS TOOL")
    print("==============================")

    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Transpose")
    print("5. Determinant")
    print("6. Exit")

    choice = input("Enter Choice : ")

    if choice == "1":

        A = input_matrix("Matrix A")
        B = input_matrix("Matrix B")

        if A.shape == B.shape:
            print("\nResult")
            print(A + B)
        else:
            print("Matrix dimensions must be same.")

    elif choice == "2":

        A = input_matrix("Matrix A")
        B = input_matrix("Matrix B")

        if A.shape == B.shape:
            print("\nResult")
            print(A - B)
        else:
            print("Matrix dimensions must be same.")

    elif choice == "3":

        A = input_matrix("Matrix A")
        print(A)
        B = input_matrix("Matrix B")
        print(B)

        if A.shape[1] == B.shape[0]:
            print("\nResult")
            print(np.matmul(A, B))
        else:
            print("Matrix multiplication not possible.")

    elif choice == "4":

        A = input_matrix("Matrix")

        print("\nTranspose")
        print(A.T)

    elif choice == "5":

        A = input_matrix("Matrix")

        if A.shape[0] == A.shape[1]:

            print("\nDeterminant")
            print(np.linalg.det(A))

        else:
            print("Determinant exists only for square matrices.")

    elif choice == "6":

        print("Thank You!")
        break

    else:

        print("Invalid Choice")

except KeyboardInterrupt:
      print("You Pressed a Key to Exit Program:")
```
