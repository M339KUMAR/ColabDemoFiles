```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = '/content/sample_data/Palo Alto Networks.xlsx'
```

```python
df = pd.read_excel(path)
df.sample(1).T
```

```python
df.info()
```

```python
df.describe(include='object').T
```

```python
df.describe(include='number').T
```

```python
df.select_dtypes(include='number').columns
```

```python
for col in df.select_dtypes('number').columns:
    print()
    print(f'Value Counts in :: {col} of type :: {type(col)}')
    print(df[col].value_counts().sort_values())
    print(f'Minimum:{df[col].value_counts().min()} & Maximum:{df[col].value_counts().max()}')
    print(f'Minimum:{df[col].min()} & Maximum:{df[col].max()}')
    print(f'Shape of the Column:{df[col].shape}')
    print(f'Data Type of {col}:: {df[col].dtype}')
    print('----------------------')

print()
print('Total Number of Int & Float Columns::',df.select_dtypes('number').columns.shape[0])
```

```python
for col in df.select_dtypes('object').columns:
    print(f'Value Counts in :: {col}')
    print('----------------------')
    print(df[col].value_counts())
    print('----------------------')

print()
print(f'Total number of Object colmns:{df.select_dtypes('object').columns.shape[0]}')
```

```python
Num_Col = df.select_dtypes(include='number').columns
Cat_Col = df.select_dtypes(include='object').columns
```

```python
print("       Numerical Columns           ")
print("----------------------------")
display(Num_Col)
print("----------------------------")
print("       Categorical Columns         ")
print("----------------------------")
display(Cat_Col)
```
